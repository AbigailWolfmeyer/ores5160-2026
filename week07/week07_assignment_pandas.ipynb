```python
# Replace infinite updated data with nan using NumPy
airquality_from_file.replace([np.inf, -np.inf], np.nan, inplace=True)

# Drop rows with NaN
airquality_from_file.dropna(inplace=True)

# Look at the results
airquality_from_file.describe(include='all')
```

Q3. Data Visualization

	1.	Create a histogram of the pm25 values.\\

	2.	Create a boxplot for the pm25 column.\\


```python
import matplotlib.pyplot as plt
```

```python
airquality_from_file[airquality_from_file["pollutant"] == "pm25"]["value"].hist()

plt.xlabel("PM2.5 Value")
plt.ylabel("Frequency")
plt.title("Histogram of PM2.5 Values")
plt.show()
```

```python
airquality_from_file[airquality_from_file["pollutant"] == "pm25"].boxplot(column="value")

plt.ylabel("PM2.5 Value")
plt.title("Boxplot of PM2.5 Values")
plt.show()
```
